```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("../data/ChargingRecords.csv")

print("Dataset loaded successfully")
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data.dtypes
```

Remove duplicate records

```python
data.drop_duplicates(inplace=True)
```

```python
print(data.duplicated().sum())
```

missing value

```python
F=data.isnull().sum()
print(F/len(data)*100)
```

```python
data["StartDatetime"] = pd.to_datetime(
    data["StartDatetime"],
    errors="coerce"
)

data["EndDatetime"] = pd.to_datetime(
    data["EndDatetime"],
    errors="coerce"
)
```

```python
data.to_csv("../data/ChargingRecords_cleaned.csv", index=False)
print("Cleaned dataset saved successfully")
```
